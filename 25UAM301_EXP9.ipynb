```python
import numpy as np
import pandas as pd

# Simple dataset
df = pd.DataFrame({'X': [1, 2, 8, 9]})

print("Dataset:")
print(df)

# Enter K
k = int(input("\nEnter the value of K: "))

# Initial centroids: first K values
centroids = df['X'].values[:k].astype(float)

iteration = 1

while True:
    print("\nIteration:", iteration)
    print("Centroids:", centroids)

    # Calculate absolute distance
    distances = np.abs(
        df['X'].values[:, np.newaxis] - centroids
    )

    # Assign points to nearest centroid
    labels = np.argmin(distances, axis=1)
    df['Cluster'] = labels + 1

    print("\nCluster Assignment:")
    print(df)

    # Calculate arithmetic mean of each cluster
    new_centroids = np.array([
        df['X'].values[labels == i].mean()
        for i in range(k)
    ])

    print("\nUpdated Centroids:", new_centroids)

    # Stop when centroids do not change
    if np.allclose(centroids, new_centroids):
        print("\nCentroids converged!")
        break

    centroids = new_centroids
    iteration += 1

# Display final clusters
print("\nFinal Clusters:")

for i in range(k):
    print("Cluster", i + 1, ":",
          df[df['Cluster'] == i + 1]['X'].tolist())
```
